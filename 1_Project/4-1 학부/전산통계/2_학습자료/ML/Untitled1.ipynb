```python
import numpy as np
x = np.array([1,2,3])
y = np.array([1,3,2])

xbar = np.mean(x)
ybar = np.mean(y)

xbar, ybar
```

```python
n = x.shape[0]
sumxx = sumxy = 0
for i in range(n):
    sumxx += (x[i] - xbar)**2
    sumxy += (x[i] - xbar) * (y[i] - ybar)

beta = sumxy / sumxx
alpha = ybar - beta*xbar

alpha, beta
```

```python
X = np.array([[1,1], [1,2], [1,3]])
Y = np.array([1,3,2])

_tmp1 = np.linalg.inv(np.dot(X.T, X))
_tmp2 = np.dot(X.T, Y)

beta = np.dot(_tmp1, _tmp2)
beta
```
