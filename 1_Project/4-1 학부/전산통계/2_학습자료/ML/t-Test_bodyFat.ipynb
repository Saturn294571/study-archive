```python
datas = ['m', 13.3, 'f', 22, 'm', 19, 'f', 26, 'm', 20,'f', 16, 'm', 8, 'f', 12, 'm', 18, 'f', 21.7,'m', 22,
        'f', 23.2, 'm', 20, 'f', 21, 'm', 31,'f', 28, 'm', 21, 'f', 30, 'm', 12, 'f', 23,'m', 16, 'm', 12, 'm', 24]
```

```python
sex = []
fat = []
for val in datas:
    if type(val) == str:
        sex.append(val)
    else:
        fat.append(val)
```

```python
fat
```

```python
import pandas as pd
dict = {'sex': sex, 'fat': fat}
df = pd.DataFrame(dict)
df
```

```python
groupby_gender = df.groupby('sex') # pandas object
```

```python
groupby_gender.mean(), groupby_gender.std(), groupby_gender.count()
```

```python
groupby_gender.boxplot(column=['fat'])
```

```python
from scipy import stats
female_fat = df[df['sex'] == 'f']['fat']
male_fat = df[df['sex'] == 'm']['fat']
stats.ttest_ind(female_fat, male_fat, equal_var=True)
```
