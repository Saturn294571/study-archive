#### 이진탐색트리 ADT
    
* root = BNode(), BNode는 item과 left, right 를 가지는 객체다.
* search method: root 부터 item이 root값보다 작으면 왼쪽으로, 크면 오른쪽으로 이동하면서 탐색한다.
   탐색에 성공하면 해당노드를 리턴하고 실패하면 None을 리턴한다.<br>
    
* insert method: 삽입은 탐색한 후, 실패한 위치에 삽입한다.
* inOrder method: 중위탐색을 실행하여 트리로부터 정렬된 결과를 리턴한다.
* minimum method: 트리의 최소값을 찾는다. 최소값은 왼쪽으로 내려 가다가 None을 만났을 때 부모노드의 값이다.
* delMin method: 트리의 최소노드를 삭제한다.
    
* delete Method: 특정 값을 가지는 노드를 삭제한다. <br> 삭제연산은 세가지 경우가 존재한다.

#### 재귀를 이용한 이진탐색트리 구현 알고리즘

  
* 삽입 알고리즘
    
  * 아래 트리에서35를 삽입한다고 하자. <br>
    step 1. curNode = 50에서 item = 35 는50보다 작으므로 curNode.left = insert(curNode.left, 35)<br>
    step 2. curNode = 30에서 item = 35 는 30보다 크므로 curNode.right = insert(curNode.right, 35)<br>
    step 3. curNode = 40에서 item = 35 는 40보다 작으므로 curNode.left = insert(curNode.left, 35)<br>
    step 4. curNode = None이므로  return node(item = 35)<br>
    step 5. step 3의 curNode 40의 curNode.left = insert(curNode.left, 35) = 35노드 & return curNode = 40<br>
    step 6. step 2의 curNode 30의 curNode.right = insert(curNode.right, 35) = 40 노드 & return curNode = 30<br>
    step 7. step 1의 curNode 50의 curNode.left = insert(curNode.left, 35) = 30 노드 & return curNode = 50<br>
    
    
  <img src="https://drive.google.com/uc?id=1oPtxAZszy23QbTrSxdTYtLOajQOJLgLt" width="500" height="300" ></center>     
   
    
    
    

```
class BNode:
    def __init__(self, item, left=None, right=None):
        self.item = item
        self.left = left
        self.right = right

class BSTree:
    def __init__(self):
        self.root = None

    def insert(self, item):  # item을 루트에서 삽입할 위치를 찾는다.
        self.root = self._insert(self.root, item)

    def _insert(self, curNode, item):
        # curNode 부터 밑으로 item 삽입위치를 찾아 삽입한다.
        # 현재 노드가 None이면 현재노드에 새노드를 삽입한다.
        if curNode == None:
            curNode = BNode(item)
        elif item < curNode.item:
        # 삽입할 값이 현재 노드값 보다 작으면 좌측을 현재노드로 보고 삽입을 실행한다.
            curNode.left = self._insert(curNode.left, item)
        else:
        # 삽입할 값이 현재 노드값 보다 크면 우측을 현재노드로 보고 삽입을 실행한다.
            curNode.right = self._insert(curNode.right, item)
        # 재귀가 끝나서 새노드를 삽입했다면 자신을 호출한 프로세스에 새노드를 리턴한다.
        return curNode

    def inOrder(self):
        return self._inOrder(self.root)

    def _inOrder(self, node, sorted=[]):
        # 현재 노드가 비어있지 않으면 일단, 계속 좌측으로 간다. 좌측의 끝에 도달하면 노드를 출력하고 오른쪽으로 이동한다.
        if node is not None:
            self._inOrder(node.left, sorted) # 좌로 이동
            sorted.append(node.item)         # 좌에 None이면 돌아오면서 curNode를 append, print 대신에 프린트 순서대로 append 하는 것임
            self._inOrder(node.right, sorted)
        return sorted

    def levelOrder(self, node=None):
        if node == None:
            node = self.root
        h = self.height(node) # 트리의 높이를 구한다.
        for i in range(1, h + 1): # i = 1층, 2층, ...
            self._levelOrder(node, i)

    def _levelOrder(self, node, level): # 해당 node에서 level = i 인 높이에 있는 노드를 출력한다.
        if node is None:
            return
        if level == 1:                                # level = 1 이면 목적지에 온 것이다.
            print(node.item, end=" ")
        elif level > 1:
            self._levelOrder(node.left, level - 1)    # level을 줄이면서 좌측 끝으로 간다.
            self._levelOrder(node.right, level - 1)   # level을 줄이면서 우측 끝으로 간다.

    def height(self, node):
        if node is None:
            return 0
        else:
            # Compute the height of each subtree
            lheight = self.height(node.left)
            rheight = self.height(node.right)

            # Use the larger one
            if lheight > rheight:
                return lheight + 1
            else:
                return rheight + 1


a = [50, 30, 80, 10, 40, 90]
t = BSTree()
for val in a:
    t.insert(val)
t.insert(35)
print(t.inOrder())
print(t.levelOrder())
```

* 탐색 알고리즘
  * 아래 트리에서 40을 탐색한다고 하자.<br>
  <center>
  <img src=" https://drive.google.com/uc?id=1nqJDheChedc-xl7G0wzixOXN-u_EU45f" width="500" height="300" ></center>  

   step 1: 루트부터 아이템을 찾는다. return self._search(self.root, item) <br>
   step 2: curNode = self.root  <br>
   만약, curNode.item > item 이면 왼쪽으로 curNode를 바꾸고 반대로  curNode.item < item 이면 오른쪽으로 curNode를 바꾸고 찾는다. <br>
   움직이면서   curNode.item == item 으로 노드를 찾으면 해당 노드를 리턴한다.
   curNode == None 이면 찾는 노드가 없는 것이므로 에러 메세지를 출력한다.<br>
   step 3: 움직일 때 마다 return 문에 의해 자신을 호출한 재귀프로세스에 최종적으로 찾은 item을 리턴해야 한다. 리턴이 없으면 찾아도 리턴되지 않으므로 재귀를 끝낼때 찾은 노드를 확인할 수 없다.
   
  

```
class BNode:
    def __init__(self, item):
        self.item = item
        self.left = None
        self.right = None

    def setLeft(self, node):
        self.left = node

    def setRight(self, node):
        self.right = node


class BSTree:
    def __init__(self, root):
        self.root = root

    def insert(self, item):  # item을 삽입한다.
        self.root = self._insert(self.root, item)

    def _insert(self, curNode, item):
        # curNode 부터 밑으로 item 삽입위치를 찾아 삽입한다.
        # 현재 노드가 None이면 현재노드에 새노드를 삽입한다.
        if curNode == None:
            curNode = BNode(item)
        elif item < curNode.item:
        # 삽입할 값이 현재 노드값 보다 작으면 좌측을 현재노드로 보고 삽입을 실행한다.
            curNode.left = self._insert(curNode.left, item)
        else:
        # 삽입할 값이 현재 노드값 보다 크면 우측을 현재노드로 보고 삽입을 실행한다.
            curNode.right = self._insert(curNode.right, item)
        # 재귀가 끝나서 새노드를 삽입했다면 자신을 호출한 프로세스에 새노드를 리턴한다.
        return curNode

    def search(self, item):
        # 루트부터 item을 찾는다.
        return self._search(self.root, item)

    def _search(self, curNode, item):
        # curNode 밑에서 item을 찾는다.
        if curNode == None:
            print("item not found")
        elif curNode.item == item:
            # 찾은 item을 리턴한다.
            return curNode
        elif item < curNode.item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            # 리턴이 없으면 찾아도 리턴되지 않으므로 재귀를 끝낼때
            # 리턴에 의해 item을 연쇄적으로 반환해야 한다.
            return self._search(curNode.left, item)
        elif item > curNode.item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            return self._search(curNode.right, item)

    def inOrder(self):
        return self._inOrder(self.root, [])

    def _inOrder(self, node, sorted=[]):
        # 현재 노드가 비어있지 않으면 일단, 계속 좌측으로 간다.
        # 좌측의 끝에 도달하면 노드를 출력하고 오른쪽으로 이동한다.
        if node is not None:
            self._inOrder(node.left, sorted)
            sorted.append(node.item)
            self._inOrder(node.right, sorted)
        return sorted


a50 = BNode(50)
a30 = BNode(30)
a80 = BNode(80)
a10 = BNode(10)
a40 = BNode(40)
a90 = BNode(90)

a50.left = a30
a50.right = a80
a80.right = a90
a30.left = a10
a30.right = a40

t = BSTree(a50)
t.insert(35)
print(t.search(35).item)


```

* 트리 최소값 찾기 알고리즘
  * 루트 부터 왼쪽 노드로 이동한다. 이동하는 도중에 None을 만나면 None의 부모노드가 최소값을 가진 노드이므로 부모노드 값을 리턴한다.

```
class BSTree:
    def __init__(self, root):
        self.root = root

    def insert(self, item):  # item을 삽입한다.
        self.root = self._insert(self.root, item)

    def _insert(self, curNode, item):
        # curNode 부터 밑으로 item 삽입위치를 찾아 삽입한다.
        # 현재 노드가 None이면 현재노드에 새노드를 삽입한다.
        if curNode == None:
            curNode = BNode(item)
        elif item < curNode.item:
        # 삽입할 값이 현재 노드값 보다 작으면 좌측을 현재노드로 보고 삽입을 실행한다.
            curNode.left = self._insert(curNode.left, item)
        else:
        # 삽입할 값이 현재 노드값 보다 크면 우측을 현재노드로 보고 삽입을 실행한다.
            curNode.right = self._insert(curNode.right, item)
        # 재귀가 끝나서 새노드를 삽입했다면 자신을 호출한 프로세스에 새노드를 리턴한다.
        return curNode

    def search(self, item):
        # 루트부터 item을 찾는다.
        return self._search(self.root, item)

    def _search(self, curNode, item):
        # curNode 밑에서 item을 찾는다.
        if curNode == None:
            print("item not found")
        elif curNode.item == item:
            # 찾은 item을 리턴한다.
            return curNode
        elif curNode.item > item:
            # 최종적으로 찾은 item을 리턴해야 한다. 리턴이 없으면 찾아도 리턴되지 않으므로 재귀를 끝낼때 리턴에 의해 item을 연쇄적으로 반환해야 한다.
            return self._search(curNode.left, item)
        elif curNode.item < item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            return self._search(curNode.right, item)

    def minimum(self):
        # 트리의 최소값을 찾는다. 최소값은 왼쪽 None을 만났을 때 부모노드의 값이다.
        return self._minimum(self.root.left)

    def _minimum(self, node):
        if node.left != None:
            return self._minimum(node.left)
        else:
            return node

    def inOrder(self):
        return self._inOrder(self.root, [])

    def _inOrder(self, node, sorted=[]):
        # 현재 노드가 비어있지 않으면 일단, 계속 좌측으로 간다. 좌측의 끝에 도달하면 노드를 출력하고 오른쪽으로 이동한다.
        if node is not None:
            self._inOrder(node.left, sorted)
            sorted.append(node.item)
            self._inOrder(node.right, sorted)
        return sorted


a50 = BNode(50)
a30 = BNode(30)
a80 = BNode(80)
a10 = BNode(10)
a40 = BNode(40)
a90 = BNode(90)

a50.left = a30
a50.right = a80
a80.right = a90
a30.left = a10
a30.right = a40

t = BSTree(a50)
t.insert(35)
print(t.search(40).item)
print(t.minimum().item)

```

* 최소노드 삭제 알고리즘
  * 아래 그림처럼 왼쪽으로 내려가면서 curNode.left가 None이면 curNode.right를 리턴하여 curNode에 저장한다.
  * 최소노드는 가장 좌측 노드가 None일 때, 해당 노드의 부모노드가 최소 노드다.
  <img src=" https://drive.google.com/uc?id=1ToyFS9wRlyzZWXIu0RVkqs40-YJww1gO" width="500" height="300" ></center>  


```
class BSTree:
    def __init__(self, root):
        self.root = root

    def insert(self, item):  # item을 삽입한다.
        self.root = self._insert(self.root, item)

    def _insert(self, curNode, item):
        # curNode 부터 밑으로 item 삽입위치를 찾아 삽입한다.
        # 현재 노드가 None이면 현재노드에 새노드를 삽입한다.
        if curNode == None:
            curNode = BNode(item)
        elif item < curNode.item:
        # 삽입할 값이 현재 노드값 보다 작으면 좌측을 현재노드로 보고 삽입을 실행한다.
            curNode.left = self._insert(curNode.left, item)
        else:
        # 삽입할 값이 현재 노드값 보다 크면 우측을 현재노드로 보고 삽입을 실행한다.
            curNode.right = self._insert(curNode.right, item)
        # 재귀가 끝나서 새노드를 삽입했다면 자신을 호출한 프로세스에 새노드를 리턴한다.
        return curNode

    def search(self, item):
        # 루트부터 item을 찾는다.
        return self._search(self.root, item)

    def _search(self, curNode, item):
        # curNode 밑에서 item을 찾는다.
        if curNode == None:
            print("item not found")
        elif curNode.item == item:
            # 찾은 item을 리턴한다.
            return curNode
        elif item < curNode.item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            # 리턴이 없으면 찾아도 리턴되지 않으므로 재귀를 끝낼때
            # 리턴에 의해 item을 연쇄적으로 반환해야 한다.
            return self._search(curNode.left, item)
        elif item > curNode.item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            return self._search(curNode.right, item)

    def minimum(self):
        # 트리의 최소값을 찾는다.
        # 최소값은 왼쪽 None을 만났을 때 부모노드의 값이다.
        return self._minimum(self.root.left)

    def _minimum(self, node):
        if node.left != None:
            return self._minimum(node.left)
        else:
            return node

    def delMin(self):
        # 최소값을 가진 노드를 삭제한다.
        self.root = self._delMin(self.root)

    def _delMin(self, node):
        # 현재 노드 좌측이 None면, 현재노드가 최소값이고
        # 현재 노드 우측값이 현재노드를 제거했을 때 최소값이므로
        # 부모노드의 왼쪽에 연결한다.(p 153 그림 5-10)
        # 최소노드의 자식노드가 모두 None이라면 부모노드의
        # 왼쪽에 None이 리턴된다.
        if node.left == None:
            return node.right
        node.left = self._delMin(node.left)
        return node

    def inOrder(self):
        return self._inOrder(self.root, [])

    def _inOrder(self, node, sorted=[]):
        # 현재 노드가 비어있지 않으면 일단, 계속 좌측으로 간다.
        # 좌측의 끝에 도달하면 노드를 출력하고 오른쪽으로 이동한다.
        if node is not None:
            self._inOrder(node.left, sorted)
            sorted.append(node.item)
            self._inOrder(node.right, sorted)
        return sorted


a50 = BNode(50)
a30 = BNode(30)
a80 = BNode(80)
a10 = BNode(10)
a40 = BNode(40)
a90 = BNode(90)

a50.left = a30
a50.right = a80
a80.right = a90
a30.left = a10
a30.right = a40

t = BSTree(a50)
t.insert(35)
print(t.inOrder())
t.delMin()
print(t.inOrder())

```

```
class BNode:
    def __init__(self, item):
        self.item = item
        self.left = None
        self.right = None

    def setLeft(self, node):
        self.left = node

    def setRight(self, node):
        self.right = node

class BSTree:
    def __init__(self, root):
        self.root = root

    def insert(self, item):  # item을 삽입한다.
        self.root = self._insert(self.root, item)

    def _insert(self, curNode, item):
        # curNode 부터 밑으로 item 삽입위치를 찾아 삽입한다.
        # 현재 노드가 None이면 현재노드에 새노드를 삽입한다.
        if curNode == None:
            curNode = BNode(item)
        elif item < curNode.item:
        # 삽입할 값이 현재 노드값 보다 작으면 좌측을 현재노드로 보고 삽입을 실행한다.
            curNode.left = self._insert(curNode.left, item)
        else:
        # 삽입할 값이 현재 노드값 보다 크면 우측을 현재노드로 보고 삽입을 실행한다.
            curNode.right = self._insert(curNode.right, item)
        # 재귀가 끝나서 새노드를 삽입했다면 자신을 호출한 프로세스에 새노드를 리턴한다.
        return curNode

    def search(self, item):
        # 루트부터 item을 찾는다.
        return self._search(self.root, item)

    def _search(self, curNode, item):
        # curNode 밑에서 item을 찾는다.
        if curNode == None:
            print("item not found")
        elif curNode.item == item:
            # 찾은 item을 리턴한다.
            return curNode
        elif curNode.item > item:
            # 최종적으로 찾은 item을 리턴해야 한다. 리턴이 없으면 찾아도 리턴되지 않으므로 재귀를 끝낼때 리턴에 의해 item을 연쇄적으로 반환해야 한다.
            return self._search(curNode.left, item)
        elif curNode.item < item:
            # 최종적으로 찾은 item을 리턴해야 한다.
            return self._search(curNode.right, item)

    def inOrder(self):
        return self._inOrder(self.root, [])

    def _inOrder(self, node, sorted=[]):
        # 현재 노드가 비어있지 않으면 일단, 계속 좌측으로 간다. 좌측의 끝에 도달하면 노드를 출력하고 오른쪽으로 이동한다.
        if node is not None:
            self._inOrder(node.left, sorted)
            sorted.append(node.item)
            self._inOrder(node.right, sorted)
        return sorted

    def minimum(self):
        # 트리의 최소값을 찾는다. 최소값은 왼쪽 None을 만났을 때 부모노드의 값이다.
        return self._minimum(self.root.left, self.root)

    def _minimum(self, node):
        if node.left != None:
            return self._minimum(node.left)
        else:
            return node

    def delMin(self):
        # 최소값을 가진 노드를 삭제한다.
        #
        self.root = self._delMin(self.root)

    def _delMin(self, node):
        # 현재 노드 좌측이 None면, 현재노드가 최소값이고 현재 노드 우측값이 현재노드를 제거했을 때 최소값이므로 부모노드의 왼쪽에 리턴한다.(p 152 그림 5-9, 5-10)
        # 만약 리턴된 node.right 역시 None이라면 즉, 최소노드의 자식노드가 모두 None이라면 부모노드의 왼쪽에 None이 리턴된다.
        if node.left == None:
            return node.right
        node.left = self._delMin(node.left)
        return node

    def delete(self, item):
        self.root = self._delete(self.root, item)

    def _delete(self, node, item):
        if node == None: # 못 찾으면 None을 리턴한다.
            return None
        if item < node.item: # 삭제노드가 왼쪽에 있다면
            node.left = self._delete(node.left, item)
        elif item > node.item: # 삭제노드가 오른쪽에 있다면
            node.right = self._delete(node.right, item)
        else: # 삭제노드를 찾았다면
            if node.right == None: # 자식이 1명 이하인 경우
                return node.left # 삭제노드 위치에 삭제노드의 좌측노드를 리턴한다.(좌측노드가 비어 있을 수도 있음)
            elif node.left == None:
                return node.right  # 삭제노드 위치에 삭제노드의 우측노드를 리턴한다.(우측노드는 비어 있지 않음)
            # 삭제노드의 자식이 둘 다 있다면 삭제 노드 우측에서 최소노드를 후계자로 지정하고
            # 밑에 있는 후계자노드는 삭제한다.
            delNode = node
            node = self._minimum(delNode.right) # 삭제 노드 오른쪽에서 후계자 노드를 찾아 지정, node가 후계자다.
            node.right = self._delMin(delNode.right) # 삭제 노드 오른쪽에서 후계자를 삭제, 삭제된 하위 트리가 후계노드 오른쪽에 연결한다.
            node.left = delNode.left # 삭제노드 좌측 트리를 후계노드 좌측으로 연결한다.
        return node

import numpy as np
x = [60, 50, 70, 20, 10, 45, 35, 25, 30, 40]
print(np.median(x)) # 37.5

x = [38] + x[:-1]
print(x)
```

#### Red Black Tree

* Red Black 트리는 기본적으로 균형 이진탐색트리다.
*  Red-Black Tree는 탐색, 삽입, 삭제 모두 시간복잡도가 log(n) 이다.
* 일반적으로 이진탐색트리는 많이 사용된다. 하지만, 연산 속도차이가 나면 안되는 경우, 균형 이진 탐색트리로 AVL, Red-Black Tree를 사용할 수 있는데 이 때, AVL 보다는 Red-Black Tree가  많이 사용된다.
* 하지만, 프로그램 구조가 복잡하다는 단점이 있다.
* 탐색은 이진탐색트리 알고리즘과 동일하고 삽입, 삭제시 트리의 높이를 log(n)이 되도록 재배열하는 작업을 수행한다.
* Red Black Tree는 아래의 조건을 만족한다.
  1. 각 노드는 레드 혹은 블랙 이다.
  2. 루트노드는 블랙이다.
  3. 모든 NIL 노드는 블랙이다. (NIL은 None을 의미함)
  4. 레드 노드의 자식은 블랙이다. (double red condition)
  5. 모든 노드에 대해 해당 노드부터 잎 노드까지 경로에 존재하는 블랙노드 수는 같다.
  
        ![]( https://drive.google.com/uc?id=1p6O--rUTPlmwZAy-Fsg5P9HQDsvGPtPw) </center>

* Red Black Tree의 특정 노드의 높이는 2 log(n+1) 보다 작다.

```
import bintrees # PC에서 실습

x = [(13,'a13'),(8,'a8'), (17,'a17'), (1,'a1'), (11,'a11'), (15,'a15'), (25,'a25'), (6,'a6'), (22,'a22'), (27,'a27')]

t = bintrees.RBTree(x)

print(t._root.value)
print(t._root.left.key, t._root.left.value)
print(t.get_value(25))
print(t.min_item(),t.max_item())

t.insert(0,'a0')
print(t.min_item(),t.max_item())
t.remove(0)
print(t.min_item(),t.max_item())
```

#### KDTree
  * KD Tree는 k dimensional Tree의 약어로 k 차원의 데이터를 빠르게 탐색할 수 있는 자료구조다.
  * kNN, k means clustering 알고리즘은 나와 가장 가까운 개체를 찾아야 한다.
  * 나와 가장 가까운 개체를 찾으려면 모든 개체와의 거리를 계산한 후, 그 중 가장 가까운 개체를 찾아야 하는데 개체수가 증가하면 계산량이 기하급수적으로 증가한다.
  * 이러한 경우, KD Tree를 이용하면 빠르게 찾을 수 있다.
  * 아래는 2차원 KDTree다. 첫번째 축은 X 축, 두번째 축은 Y축 순서로 진행한다.
  * (7,2), (5,4), (9,6), (2,3), (4,7), (8,1) 순서로 입력된다고 가정할 때
  * (5,4)는 x값 5가 7보다 작으므로 (7,2)의 좌측에 저장된다.
  *(9,6)은 x값 9가 7보다 크므로 (7,2)의 우측에 저장된다.
  *(2,3)은 x값 2가 7보다 작고 y값 3이 두번째 노드(5, 4)보다 작으므로 좌측에 저장된다.
  * 이러한 방식으로 모든 데이터가 저장된다.
  * 트리의 루트는 X 축의 중앙값, Y 축의 중앙값에 가까운 개체를선택하는 것이 좋다.
