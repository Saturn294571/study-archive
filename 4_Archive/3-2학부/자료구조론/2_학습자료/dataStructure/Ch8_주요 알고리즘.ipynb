

### 분할정복(Divide and Conquer)    
* 분할정복은 문제를 두개 이상의 조각으로 분할하여 각 분할을 해결하고 이를 결합하여 전체 문제를 해결하는 방식이다.

* 예를 들어, 아래 토지를 손실없이 몇 개의 정사각형 모양으로 나누어 보자.

알고리즘은 아래와 같다.
위의 토지를 정사각형과 정사각형이 아닌 부분으로 분할한 다음, 정사각형이 아닌 부분에 대해
다시 분할정복 알고리즘을 적용한다. <br>
즉, 직사각형을 정사각형과 직사각형의 합으로 만들고 작은 직사각형을 다시 정사각형과 직사각형의 합으로 만드는 과정을 반복하는 분할 정복 방식의 문제 해결이다.

아래 알고리즘을 수행하면 가장 큰 정사각형은 80 * 80 이다.

사실, 이 알고리즘은 1680, 640의 최대공약수를 구하는 것이다.

```
# 위의 과정을 그대로 프로그래밍 해보자.

a = 1680        # 큰 수
b = 640         # 작은 수

while b > 0:
    print(a, b)
    r = a % b
    a = b
    b = r

print(a, b)
```

### Euclid Algorithm
gcd(a,b) = gcd(b, a% b), gcd: greatest common divisor



```
# divide and conquer 방식으로 구현한 알고리즘
# a = b, b = a % b 로 업데이트 한다.
# Euclid Algorithm: gcd(a,b) = gcd(b, a%b)
def gcd(a,b):
    if b == 0: return a
    return gcd(b,a % b)

gcd(1680, 640)
```

### Set Covering Problem

* 미국의 주는 50개인데 50개 주 전체에 방송을 해야 한다.
* 하나의 라디오 방송국이  하나 이상의 주를 커버할 수 있다고 가정할 때, 50개 주를 다 커버하도록 하면서도 최소의 라디오 방송국을 방문하려면 방송국을 어떻게 방문해야 하는가?
* 시행착오법을 통해 모든 가능한 경우의 수를 시행해야 알 수 있다.

<center><center><img src='https://drive.google.com/uc?id=1a-hvN6JhVLtPvL7UyCz31dTufD7CWdH4' width='500' height='300' /></center></center>

* 일단, 주에 대한 방문여부를 나타내는 모든 가능한 경우의 수를 생각해 보자.
   모든 가능한 경우의 수는  2^50 개 존재한다. <br>
   우리는 이것을 멱집합(Power Set)이라고 한다.

* 위의 멱집합에서 모든 주를 커버하는 집합 중 가장 작은 방송국을 포함하는 곳을 방문하면 된다. 하지만, 매우 큰 작업이다.

*  Power Set of {a,b,c}: P(S) = { {}, {a}, {b}, {c}, {a, b}, {a, c}, {b, c}, {a, b, c} }

* Power Set Algorithm
  * 멱집합 알고리즘은 아래 그림과 같이 2진수를 이용하여 구현할 수 있다.
  * {a, b, c} 세 개를 '000' 2진수로 표현하고 1씩 증가시키면서 모든 가능한 조합을 생성한다.

    <center><center><img src='https://drive.google.com/uc?id=1QlmWyyXZ9_lsm2I1jXJPV14iy51EIMMb' width='500' height='300' /></center></center>
    

```
# Power Set Algorithm

def binAdd(a): # 이진수에 마지막 비트에 "1"을 더하는 함수
    n = len(a)
    curPos = n-1

    while True:
        a[curPos] = (a[curPos] + 1) % 2
        if a[curPos] == 0:
            curPos -= 1
        else:
            break
    return a

binAdd([0,1,0])
```

```
x = ['a','b','c']
n = len(x)
bin = [0] * n # 초기 이진수를 생성한다.

for i in range(2**n-1):
    _tmp = binAdd(bin) # 이진수에 1을 더한다.
    powerSet = []      # 이진수를 문자로 표현한다.
    for j in range(n):
        if _tmp[j] == 1:
            powerSet.append(x[j])
    print(_tmp, powerSet)
```

### 근사 알고리즘
* 2^50 조합에 대해 다 검사할 수 없으니 위 문제를 탐욕 알고리즘으로 최적은 아니지만, 최선의 해를 구해 보자.
* 목표는 최소한의 방문으로 50개의 주에 방송하는 것이므로 방문으로 가장 많은 주를 커버할 수 있는 방송국을 우선적으로 택하는 알고리즘을 적용한다.
* 항상, 가장 많은 주를 커버하는 방송국을 선택하는 것이 탐욕스러운 것이다.

 <br>
    Step 1: 아직 방송이 되지 않은 주를 가장 많이 포함하고 있는 방송국을 선택한다. <br>
    Step 2: 위에서 선택한 방송국을 제외하고 남은 방송국 중에 가장 많은 지역을 커버하는 방송국을 선택한다. <br>
    Step 3: 위 과정을 50개주를 모두 커버할 때 까지 반복한다.


```
states_needed = set(['mt','wa','or','id','nv','ut','ca','az'])
stations = {}
stations['one'] = set(['id','nv','ut'])
stations['two'] = set(['wa','id','mt'])
stations['three'] = set(['or','nv','ca'])
stations['four'] = set(['nv','ut'])
stations['five'] = set(['ca','az'])
```

```
# 방송국이 모든 주를 포함하고 있는가? 즉, 방송국을 방문하면 모든 주에 방송할 수 있는지를 테스트함

stationSet = set()

for key in stations.keys():
    stationSet = stationSet | stations[key] # |: 합집합 연산자

print(stationSet)
print(stationSet == states_needed)
```

```
visits = [] # 방문한 주의 집합

while len(states_needed) > 0:
    states_covered = set()
    for station, states_for_station in stations.items(): # for looping 하면서 방문하지 않은 주를 가장 많이 커버할 수 있는 방송국을 찾는다.
        covers = states_for_station & states_needed      # 방송국별로 커버할 수 있는 주의 집합과 방송을 해야하는 주의 교집합을 구한다.
        if len(covers) > len(states_covered): # 교집합 중 가장 많은 주를 커버하는 방송국을 찾는다.
            best_station = station            # 그 방송국을 방문하면서 커버되는 주를 방문 필요 주 집합에서 제거한다.
            states_covered = covers
    states_needed = states_needed - states_covered
    visits.append(best_station)               # 방문한 주에 현재 선택한 주를 추가한다.
print(visits)
```

```
# 동전 거슬러주기 알고리즘(1원, 5원, 10원, 50원, 100원, 500원 동전을 이용)

coins = [500,100,50,10,5,1]
x = int(input("금액을 입력하세요: "))
return_coins= []
remain = x

while remain > 0:
    for coin in coins:
        if coin <= remain:
            remain -= coin
            return_coins.append(coin)
            break

print(x, return_coins)
```
