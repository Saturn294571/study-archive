```python
class Stack:
    def __init__(self):
        self.s = []

    def is_empty(self):
        return len(self.s) == 0

    def push(self, item):
        self.s.append(item)

    def pop(self):
        if self.is_empty() == False:
            return self.s.pop(-1)
        else:
            return None

    def peek(self):
        if self.is_empty() == False:
            return self.s[-1]
        else:
            return None

    def size(self):
        return len(self.s)
```

```python
s = Stack()
s.push('사과')
s.push('배')
s.push('딸기')
s.push('포도')

s.s
```

```python
s.peek()
```

```python
s.size()
```

```python
string = 'abcdefg'

string[::-1]
```

```python
s = Stack()
for _s in string:
    s.push(_s)

while s.is_empty() == False:
    print(s.pop(), end='')
```

```python
def eq_check(eq):
    s = Stack()
    for _s in eq:
        if _s == '(':
            s.push(_s)
        elif _s == ')':
            s.pop()
    return s.is_empty() == True
```

```python
eq_check("2+3*(2+4*(1+2))")
```
