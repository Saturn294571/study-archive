```python
import hashlib

def fileCompare(file1, file2):
    f1 = open(file1, 'rb')
    f2 = open(file2, 'rb')
    file1 = f1.read()
    file2 = f2.read()
    f1.close()
    f2.close()
    if hashlib.sha256(file1).hexdigest() == hashlib.sha256(file2).hexdigest():
        return True
    else:
        return False

print(fileCompare("/gdrive/My Drive/dataStructure/data/subway.csv", "/gdrive/My Drive/dataStructure/data/price.csv"))
print(fileCompare("/gdrive/My Drive/dataStructure/data/price.csv", "/gdrive/My Drive/dataStructure/data/price.csv"))
```
