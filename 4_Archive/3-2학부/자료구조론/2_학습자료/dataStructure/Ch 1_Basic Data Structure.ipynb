```
def dec2bin(dec):
    bin = []
    while True:
        frac = dec % 2
        bin.append(frac)
        dec = dec // 2
        if dec == 1:
            break
    bin.append(dec)
    return bin[::-1]

def bin2dec(bin):
    dec = 0
    n = len(bin)
    for i in range(n):
        dec += bin[i] * 2**(n-i-1)
    return dec

```

```
class binary:
    @staticmethod
    def dec2bin(dec):
        bin = []
        while True:
            frac = dec % 2
            bin.append(frac)
            dec = dec // 2
            if dec == 1:
                break
        bin.append(dec)
        return bin[::-1]

    @staticmethod
    def bin2dec(bin):
        dec = 0
        n = len(bin)
        for i in range(n):
            dec += bin[i] * 2**(n-i-1)
        return dec
```

```
bin = binary.dec2bin(59)
dec = binary.bin2dec(bin)
print(bin, dec)
```

```
class binary:
    @classmethod
    def bin2dec(cls, bin):
        summ = 0
        for i in range(len(bin)):
            summ += int(bin[i])* 2**(len(bin) - i -1)
        return summ

    @classmethod
    def dec2bin(cls, dec):
        bin = []
        mok = dec
        while (mok > 1):
            mok = dec // 2
            red = dec % 2
            dec = mok
            bin.append(str(red))
        bin.append('1')
        bin = bin[::-1]
        bin = ''.join(bin)
        return bin
```

```
bin = binary.dec2bin(21)
dec = binary.bin2dec(bin)

print(bin, dec)
```

```
dec = 21
bin= []
while dec >= 1:
    mod = dec // 2
    frac = dec - 2 * mod
    bin.append(str(frac))
    dec = mod

bin = bin[::-1]
bin = ''.join(bin)
bin
```

```
class Bin_compute:
    @classmethod
    def dec2bin(cls, dec):
        bin= []
        while dec >= 1:
            mod = dec // 2
            frac = dec - 2 * mod
            bin.append(str(frac))
            dec = mod
        bin = bin[::-1]
        bin = ''.join(bin)
        return bin

    @classmethod
    def bin2dec(cls, bin):
        dec = 0
        k = len(bin)
        for i in range(k):
            dec += int(bin[k-i-1]) * 2**i
        return dec
```

```
Bin_compute.dec2bin(21)
```

```
Bin_compute.bin2dec('10101')
```

#### ADT(Abstract Data Type)

* 기초적인 자료구조 외 연결 리스트(linked list), queue, stack, Tree, Graph 등은 기본적으로 제공하지 않으므로 직접 만들어 사용하거나 오픈소스로 제공되는 외부 라이브러리를 사용한다.
* 자료구조에서는 이러한 자료구조를 직접 만들어보고 응용해보는 공부를 할 것이다.
* 자료구조의 개략적 설계도가 ADT 다.
* ADT를 구체화하면 자료구조가 완성되는 것이다.
* 자료구조를 사용하여 문제를 해결하는 것이 알고리즘이다.
