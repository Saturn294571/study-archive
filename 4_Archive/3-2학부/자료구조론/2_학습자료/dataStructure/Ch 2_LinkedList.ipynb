```
class LinkedList:
    def __init__(self):
        self.root = None
        self.n = 0

    def append(self, item):
        newNode = Node(item)
        if self.root == None:
            self.root = newNode
        else:
            curNode = self.root
            while curNode.link != None:
                curNode = curNode.link
            curNode.link = newNode
        self.n += 1

    def insert(self, idx, item):
        newNode = Node(item)
        if idx == 0:
            newNode.link = self.root.link
            self.root = newNode
        elif idx <= self.n - 1:
            curNode = self.root
            for i in range(idx-1):
                curNode = curNode.link
            newNode.link = curNode.link
            curNode.link = newNode
        else:
            print('index Error')
        self.n += 1

    def find(self, item):
        curNode = self.root
        idx = 0
        while curNode.link != None:
            if curNode.item == item:
                return idx
            curNode = curNode.link
            idx += 1
        return -1

    def print(self):
        curNode = self.root
        while curNode.link != None:
            print(curNode.item, end = ' -> ')
            curNode = curNode.link
        print(curNode.item)
```

* Linked List 설계
 * root: 리스트 루트 노드이고 Linked List는 root 노드의 주소를 가진다.
   * root = Node()
 * append method: item을 주면 item 노드를 Linked List 마지막 노드에 추가하는 메소드
   * item으로 newNode를 만들고 root의 item이 None이면 newNode를 root로 지정하고 아니면 리스트의 끝까지 이동한 후, 마지막 노드의 링크에 newNode의 주소를 삽입
 * insert method: Linked List에 index로 새 노드를 추가하는 메소드로 index=0 이면 root로 추가하고, index 값에 따라 해당 위치에 삽입한다.
   * index < 0 or index >= listSize 이면 error 메세지를 출력한다.
   * rootNode를 0번째로 보고 index -1 번째 노드의 링크에 newNode를 넣고 newNode의 링크에      index-1 번째 링크에 있던 주소를 넣는다.
 * find Method: Linked List에서 해당 아이템을 찾고 index를 리턴한다.  없으면 -1을 리턴한다.
 * delete method: 주어진 값을 이용해 해당 노드를 찾고 지운다.
   * curNode = preNode = root로 초기 지정하고 curNode를 움직일 때 마다 preNode는 curNode의 이전 노드로 갱신한다
   * del item이 첫번째 노드면 self.root에 del item의 링크를 넣는다.
   * del item이 중간노드면preNode 링크에 del item 링크를 넣는다.
   *  마지막 노드라면 preNode 링크에 None를 넣는다.

 * listSize method: Linked List의 item 수를 리턴한다.

 * get_items method: Linked List item 전체를 리스트로 반환한다.



```
class LinkedList:
    def __init__(self):
        self.root = Node()

    def append(self, item):
        if self.root.item == None:
            self.root = Node(item)
        else:
            curNode = self.root
            newNode = Node(item)
            while curNode.link != None:
                curNode = curNode.link
            curNode.link = newNode

    def get_items(self):
        items = []
        curNode = self.root
        while curNode.link != None:
            items.append(curNode.item)
            curNode = curNode.link
        items.append(curNode.item)
        return items

    def size(self):
        curNode = self.root
        if self.root == None:
            return 0
        else:
            cnt = 1
        while curNode.link != None:
            curNode = curNode.link
            cnt += 1
        return cnt

    def search(self, item):
        idx = 0
        curNode = self.root
        while curNode.link != None:
            if curNode.item == item:
                break
            else:
                curNode = curNode.link
                idx += 1
        return idx

    def insert(self, k, item):
        newNode = Node(item)
        curNode = self.root
        if k == 0:
            _tmp = self.root
            self.root = newNode
            newNode.link = _tmp
        elif k > 0:
            for i in range(k-1):
                curNode = curNode.link
            _tmp = curNode.link
            curNode.link = newNode
            newNode.link = _tmp

    def delete(self, item):
        curNode = self.root
        idx = self.search(item)
        if idx == 0:
            self.root = self.root.link
        elif idx > 0:
            for i in range(idx-1):
                curNode = curNode.link
            curNode.link = curNode.link.link

    def pop(self):
        _tmp = self.root.item
        self.root = self.root.link
        return _tmp
```

####  다항식 덧셈 알고리즘

* Step 0: 덧셈 결과를 기록할 빈 다항식을 만든다.
* Step 1: 두개의 다항식에서 최고 차수를 찾고 최고 차수가 있는 리스트에서 팝하여 결과 다항식에 추가한다. 만약, 최고 차수를 가지는 항이 두개 다항식에 공통으로 존재하면 계수를 합하여 add한다.
* Step 2: 한쪽 다항식이 empty 할 때까지 Step 1을 수행한다.
* Step 3: 비어 있지 않은 다항식의 항을 결과 다항식에 add 한다.

```
# method 1
# 두 다항식을 더했을 때, 결과의 차수는 두 다항식 중 최대차수에 의해 결정된다.

class Poly1:
    def __init__(self, maxOrder):
        self.coef = [0] * (maxOrder + 1)
        self.maxOrder = maxOrder

    def setCoef(self, coef, order):
        self.coef[order] = coef

    def get_items(self):
        return self.coef

# 두 다항식 중 작은 차수까지 모든 차수에 대해 계수를 더하고 남은 고차항은 따로 더한다.
    @staticmethod
    def add(p, q):
        if p.maxOrder >= q.maxOrder:
            r = Poly1(p.maxOrder)
            for i in range(q.maxOrder+1):
                r.coef[i] = p.coef[i] + q.coef[i]
            for i in range(q.maxOrder+1, p.maxOrder+1):
                r.coef[i] = p.coef[i]
        else:
            r = Poly1(q.maxOrder)
            for i in range(p.maxOrder+1):
                r.coef[i] = p.coef[i] + q.coef[i]
            for i in range(p.maxOrder+1, q.maxOrder+1):
                r.coef[i] = q.coef[i]
        return r


p = Poly1(4)
p.setCoef(4,4)
p.setCoef(3,2)
p.setCoef(3,0)
print(p.get_items())

q = Poly1(3)
q.setCoef(3,3)
q.setCoef(4,2)
q.setCoef(2,1)
q.setCoef(1,0)
print(q.get_items())

r = Poly1.add(p,q)
print(r.get_items())
```

```
# method 2
class Poly2:
    def __init__(self):
        self.coef = []

    def setCoef(self,coef):
        self.coef.append(coef)

    def peek(self):
        return self.coef[0][1]

    def pop(self):
        return self.coef.pop(0)

    def size(self):
        return len(self.coef)

    def get_items(self):
        return self.coef

    @staticmethod
    def add(p, q):
        r = Poly2()
        while p.size() > 0 and q.size() > 0:
            if p.peek() > q.peek():
                r.setCoef(p.pop())
            elif p.peek() < q.peek():
                r.setCoef(q.pop())
            else:
                _tmp1 = p.pop()
                _tmp2 = q.pop()
                r.setCoef([_tmp1[0]+_tmp2[0], _tmp1[1]])

        while p.size() > 0:
            r.setCoef(p.pop())
        while q.size() > 0:
            r.setCoef(q.pop())
        return r

p = Poly2()
p.setCoef([4,4])
p.setCoef([3,2])
p.setCoef([3,0])
print(p.get_items())

q = Poly2()
q.setCoef([3,3])
q.setCoef([4,2])
q.setCoef([2,1])
q.setCoef([1,0])
print(q.get_items())

r = Poly2.add(p, q)
print(r.get_items())
```

```
# method 3
class Poly3:
    def __init__(self):
        self.parm = LinkedList()

    def setParm(self, coef, order):
        self.parm.append([coef, order])

    def peek(self):
        return self.parm.root.item[1]

    def pop(self):
        return self.parm.pop()

    def size(self):
        return self.parm.size()

    def get_poly(self):
        return self.parm.get_items()

    @staticmethod
    def add(p, q):
        r = Poly3()
        while (p.size() > 0) and (q.size() > 0):
            if p.peek() > q.peek():
                _tmp = p.pop()
                r.setParm(_tmp[0], _tmp[1])
            elif p.peek() < q.peek():
                _tmp = q.pop()
                r.setParm(_tmp[0], _tmp[1])
            else:
                _tmp1 = p.pop()
                _tmp2 = q.pop()
                r.setParm(_tmp1[0] + _tmp2[0], _tmp1[1])
        if p.size() == 0:
            while q.size() > 0:
                _tmp = q.pop()
                r.setParm(_tmp[0], _tmp[1])
        else:
            while p.size() > 0:
                _tmp = p.pop()
                r.setParm(_tmp[0], _tmp[1])

        return r

p = Poly3()
p.setParm(4,4)
p.setParm(3,2)
p.setParm(3,0)
print(p.get_poly())

q = Poly3()
q.setParm(3,3)
q.setParm(4,2)
q.setParm(2,1)
q.setParm(1,0)
print(q.get_poly())

r = Poly3.add(p, q)
print(r.get_poly())
```

### Double Linked List

* Double Linked List는 아래 그림처럼 링크가 좌, 우에 두개가 존재하는 Linked List다.
* Single Linked List는 삭제를 위해서 preNode를 관리해야 하는 불편함이 있는데 Double Linked List는 preNode를 알 수 있는 장점은 있지만 이를 위해 링크를 두 개 관리해야 하는 손실이 있다.

* Double Linked List 설계

 * DNode: item과 left, right에 각각 좌, 우 노드의 주소를 저장한다.

 * root: 리스트 루트 노드이고 Double Linked List는 root 노드의 주소를 가진다.
    * root = DNode()
    
  * append method: item을 주면 item을 가진 새 노드를 Linked List 마지막 노드에 추가하는 메소드
   * newNode를 만들고 root의 item이 None이면 newNode를 root로 지정하고 아니면 리스트의 끝까지 이동한 후, 마지막 노드의 right에 newNode의 주소를 삽입하고 newNode의 left에 마지막 노드의 주소를 삽입한다.

 * insert method: Linked List에 index로 새 노드를 추가하는 메소드로 index=0 이면 root로 추가하고, index 값에 따라 해당 위치에 삽입한다.
   * index < 0 or index >= listSize 이면 error 메세지를 출력한다.
   * rootNode를 0번째로 보고 index -1 번째 노드 주소를 newNode의 left에 넣고, index번째 노드의 주소를 newNode right에 넣는다.
   
 * find Method: Linked List에서 해당 아이템을 찾고 index를 리턴한다.  없으면 -1을 리턴한다.

 * delete method: 주어진 값을 이용해 해당 노드를 찾고 지운다.
   * del item이 첫번째 노드면 self.root에 self.root.right를 넣고 self.root.left에 None를 넣는다.
   * del item이 중간노드면 delNode.left.right에 delNode.right를 삽입한다.
   * 마지막 노드라면 마지막 노드 left.right에 None를 삽입한다.

 * listSize method: Linked List의 item 수를 리턴한다.

 * print method: Linked List를 print 한다.

```
# Node Class: item으로 value를 가지고 right, left를 가진다. link가 없을 경우, None을 입력한다.
class DNode:
    def __init__(self, item=None):
        self.item = item
        self.right = None
        self.left = None

a = DNode('사과')
print(id(a))
```

```
class DoubleLinkedList:
    def __init__(self):
        self.root = DNode()

    def append(self, item):
        newNode = DNode(item)
        if self.root.item == None:
            self.root = newNode
        else:
            curNode = self.root
            while curNode.right != None:
                curNode = curNode.right
            curNode.right = newNode
            newNode.left = curNode

    def delete(self, item):
        curNode = self.root
        if curNode.item == item:
             self.root = self.root.right
             self.root.left = None
        else:
            while curNode.right != None:
                curNode = curNode.right
                if curNode.item == item:
                    if curNode.right == None:   # 끝노드
                        curNode.left.right = None
                    else:                       # 중간노드
                        curNode.left.right = curNode.right
                        curNode.right.left = curNode.left

    def print(self):
        curNode = self.root
        while curNode.right != None:
            print(curNode.item, end = ", ")
            curNode = curNode.right
        print(curNode.item)

```

### Circle Linked List

* Circle Linked List는 아래 그림처럼 마지막 노드의 링크에 root 노드의 주소를 삽입한 것이다.
* Single Linked List는 항상 root로 부터 순차탐색을 해야 하는데 Circle Linked List는 아무 위치에서나 원형으로 탐색이 가능하다.
* 무한반복 mp3, 혹은 사진 슬라이드 쇼 플레이 순서를 만드는 예와 같이 순서가 계속 반복되는 원형 자료구조에 적합하다.

* Circle Linked List 설계
 * root: 리스트 루트 노드이고 Linked List는 root 노드의 주소를 가진다.   root = Node()
 * tail : 리스트 마지막 노드이고 tail.link = root 가 된다. 마지막 노드를 관리하는 이유는 루트가 삭제될 경우, tail.link를 갱신해야 하기 때문이다.
 * current: 리스트 순환을 위해 현재 노드를 관리한다.
  
 * append method: item을 주면 item 노드를 tail 노드 다음에 추가하는 메소드
   * newNode를 만들고 root의 item이 None이면 newNode를 root와 tail로 지정하고 newNode 링크에 root 주소를 삽입한다.  아니면 tail 링크에 newNode의 주소를 삽입하고 newNode 링크에는 root 주소를 삽입한다. 이후, newNode를 tail로 갱신한다.

 * setCurrent method: 리스트에서 item을 가지는 노드를 current 노드로 지정한다.
 * moveNext method: current 노드를 다음 노드로 바꾼다.
  * insert method: current 노드 다음에 새노드를 삽입한다.
   * current 노드 링크에 새 노드를 지정하고 새노드의 링크에 current 노드에 연결되어 있던 링크를 지정하여 연결한다.
   * 만약, current가 tail 노드 였다면 새노드를 tail 노드로 재지정한다.

 * delete method: 주어진 값을 이용해 해당 노드를 찾고 지운다.
   * curNode = preNode = root로 초기 지정하고 curNode를 움직일 때 마다 preNode는 curNode의 이전 노드로 갱신한다.
   * del item이 첫번째 노드면 root에 del item의 링크를 넣고 tail노드의 링크에 값을 root로 갱신한다.
   * del item이 중간노드면 preNode 링크에 del item의 링크를 넣는다.
   *  마지막 노드라면 preNode 링크에 del item의 링크를 넣고 tail 노드를 preNode로 갱신한다.

```
class Node:
    def __init__(self, item=None):
        self.item = item
        self.link = None

a = Node("딸기")
b = Node("바나나")
a, b
```

```
class Node:
    def __init__(self, item=None, link = None):
        self.item = item
        self.link = link

class CircleLinkedList:
    def __init__(self):
        self.root = Node()    # 시작노드
        self.tail = self.root # 끝노드
        self.current = self.root

    def append(self, item):
        # 추가할 새노드를 만든다.
        newNode = Node(item)
        # 루트가 비어 있으면 루트와 테일을 새노드로 지정한다.
        if self.root.item == None:
            self.root = newNode
            self.tail = newNode
            newNode.link = self.root
        else:
        # 루트가 비어 있지 않으면 테일 뒤에 새노드를 추가하고 테일을 업데이트한다.
            self.tail.link = newNode
            newNode.link = self.root
            self.tail = newNode

    # 리스트 안에 모든 노드를 프린트 한다.
    def pprint(self):
        curNode = self.root
        print(curNode.item)
        while curNode.link != self.root:
            curNode = curNode.link
            print(curNode.item)

    # 리스트에 존재하는 노드 수를 구한다.
    def listSize(self):
        curNode = self.root
        cnt = 1
        while curNode.link != self.root:
            curNode = curNode.link
            cnt += 1
        return cnt

    # 리스트에서 item을 가지는 노드를 current 노드로 지정한다.
    def setCurrent(self,item):
        curNode = self.root
        for i in range(self.listSize()):
            if curNode.item != item:
                curNode = curNode.link
            else:
                self.current = curNode
                break

    # current 노드를 다음 노드로 바꾼다.
    def moveNext(self):
        self.current = self.current.link
        print("현재 위치는 ", self.current.item, "입니다.")

    # current 다음에 item을 삽입한다.
    def insert(self, item):
        # 삽입할 새 노드를 만든다.
        newNode = Node(item)
        # current 노드의 링크에 새 노드를 지정하고 새노드의 링크에 current 노드가 연결되어 있던 링크를 지정하여 연결한다.
        _tmp = self.current.link
        self.current.link = newNode
        newNode.link = _tmp
        # 만약, item1이 tail 노드 였다면 새노드를 tail 노드로 재지정한다.
        if self.current == self.tail:
            self.tail = newNode

    # item을 가지는 노드를 삭제한다.
    def delete(self, item):
        delYN = False
        # 루트가 삭제 대상 노드인지 확인한다. 루트를 삭제할 경우, 루트 다음 노드를 루트로 갱신하고 tail 노드의 링크를 갱신된 루트로 재지정한다.
        curNode = self.root
        if curNode.item == item:
            self.root = self.root.link
            self.tail.link = self.root
            delYN = True
        # 루트가 삭제 대상이 아닐 경우, 끝까지 탐색하면서 item이 존재하는 노드를 찾는다.
        # 찾았다면 preNode의 링크에 대상노드 링크를 넘겨준다. 만약, 대상 노드가 tail노드라면 tail 노드를 preNode로 변경한다.
        else:
            while curNode.link != self.root:
                preNode = curNode
                curNode = curNode.link
                if curNode.item == item:
                    preNode.link = curNode.link
                    if curNode == self.tail:
                        self.tail = preNode
                    delYN = True
        # 삭제를 하지 못했으면 error message를 출력한다.
        if delYN == False: print("delete failed")

fruits = CircleLinkedList()
fruits.append('사과')
fruits.append('앵두')
fruits.append('오렌지')
fruits.append('포도')

fruits.pprint()
print(fruits.listSize())
fruits.setCurrent("포도")
fruits.insert("수박")
print(fruits.tail.item)
fruits.pprint()
```
