### 군집분석 (계층적 클러스러팅)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x1 = np.array([9,7,5,6,11,10,8,4,8,6,5,3,12,7,6,5])
x2 = np.array([8,5,3,4,9,6,5,12,7,4,2,10,8,9,4,8])
X = np.concatenate((x1.reshape(-1,1), x2.reshape(-1,1)), axis = 1)
X
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
link = linkage(X, method='complete')
plt.figure(figsize=(10,5))
dendrogram(link)
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster
predict = fcluster(link, 5, criterion='distance')
predict
```

```python
np.unique(predict, return_counts=True)
```

```python
def dendrogram_plot(P=predict):
  for i in np.unique(predict, return_counts=True)[0]:
    plt.scatter(X[P == i][:,0],X[P == i][:,1])
  plt.show()
dendrogram_plot()
```

```python
from scipy.cluster.hierarchy import fcluster
predict = fcluster(link, 4, criterion='distance')
predict
```

```python
np.unique(predict, return_counts=True)
```

```python
def dendrogram_plot(P=predict):
  for i in np.unique(predict, return_counts=True)[0]:
    plt.scatter(X[P == i][:,0],X[P == i][:,1])
  plt.show()
dendrogram_plot()
```

### K-means 비계층적 군집분석

```python
x1 = np.array([9,7,5,6,11,10,8,4,8,6,5,3,12,7,6,5])
x2 = np.array([8,5,3,4,9,6,5,12,7,4,2,10,8,9,4,8])
X = np.concatenate((x1.reshape(-1,1), x2.reshape(-1,1)), axis = 1)
X
```

```python
from sklearn.cluster import KMeans
KM = KMeans(n_clusters = 3, init='k-means++',random_state=42)
KM.fit(X)
predict = KM.labels_
predict
```

```python
KM.cluster_centers_
```

```python
def cluster_plot(P=predict):
  for i in np.unique(predict, return_counts=True)[0]:
    plt.scatter(X[P == i][:,0],X[P == i][:,1])
cluster_plot()
plt.scatter(KM.cluster_centers_[0,0], KM.cluster_centers_[0,1], c='red')
plt.scatter(KM.cluster_centers_[1,0], KM.cluster_centers_[1,1], c='red')
plt.scatter(KM.cluster_centers_[2,0], KM.cluster_centers_[2,1], c='red')
plt.show()
```

```python
print(KM.transform(X[0:1]))
KM.predict(X[0:1])
```

### 최적의 K 탐색 - 엘보우 방법

```python
#n_clusters = 3 인 경우의 이너셔
KM.inertia_
```

```python
inertia = []
for i in range(2,10):
  KM = KMeans(n_clusters = i, init='k-means++', random_state=42 )
  KM.fit(iris.data_scaled)
  inertia.append(KM.inertia_)
plt.plot(range(2,10), inertia)
plt.show()
```

```python
def visualize_cluster_plot(clusterobj, dataframe, label_name, iscenter=True):

    # 군집별 중심 위치: K-Means, Mean Shift 등
    if iscenter:
        centers = clusterobj.cluster_centers_

    # Cluster 값 종류
    unique_labels = np.unique(dataframe[label_name].values)

    markers=['o', 's', '^', 'x', '*']
    isNoise=False

    for label in unique_labels:
        # 군집별 데이터 프레임
        label_cluster = dataframe[dataframe[label_name]==label]

        if label == -1:
            cluster_legend = 'Noise'
            isNoise=True
        else:
            cluster_legend = 'Cluster '+str(label)

        # 각 군집 시각화
        plt.scatter(x=label_cluster['ftr1'], y=label_cluster['ftr2'], s=70,
                    edgecolor='k', marker=markers[label], label=cluster_legend)

        # 군집별 중심 위치 시각화
        if iscenter:
            center_x_y = centers[label]
            plt.scatter(x=center_x_y[0], y=center_x_y[1], s=250, color='white',
                        alpha=0.9, edgecolor='k', marker=markers[label])
            plt.scatter(x=center_x_y[0], y=center_x_y[1], s=70, color='k',\
                        edgecolor='k', marker='$%d$' % label)

    if isNoise:
        legend_loc='upper center'
    else:
        legend_loc='upper right'

    plt.legend(loc=legend_loc)
    plt.show()
```

```python
# k Means와 GMM 비교

from sklearn.datasets import make_blobs

# 가상데이터
X, y = make_blobs(n_samples=300, n_features=2, centers=3, cluster_std=0.5, random_state=0)

# 데이터 프레임
cluster_df = pd.DataFrame(X, columns=["ftr1","ftr2"])
cluster_df["target"] = y
# 길게 늘어난 타원형의 데이터 셋을 생성하기 위해 변환함.
transformation = [[0.60834549, -0.63667341], [-0.40887718, 0.85253229]]
X_aniso = X @ transformation

# 데이터 프레임
cluster_df = pd.DataFrame(X_aniso, columns=["ftr1","ftr2"])
cluster_df["target"] = y

visualize_cluster_plot(None, cluster_df, "target", iscenter=False)
```

```python
# KMeans
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans_label = kmeans.fit_predict(X_aniso)

# 군집 레이블 추가
cluster_df["kmeans_label"] = kmeans_label

# 군집분석 시각화
visualize_cluster_plot(kmeans, cluster_df, "kmeans_label", iscenter=True)
```

```python
# GMM
gmm = GaussianMixture(n_components=3, random_state=0)
gmm_label = gmm.fit_predict(X_aniso)

# 군집 레이블 추가
cluster_df["gmm_label"] = gmm_label

# GMM은 cluster_centers_ 속성이 없음
visualize_cluster_plot(gmm, cluster_df, 'gmm_label',iscenter=False)
```
